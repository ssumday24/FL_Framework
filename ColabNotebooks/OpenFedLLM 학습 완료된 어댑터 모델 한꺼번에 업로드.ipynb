```python
#Peft 어댑터 모델이 저장되어 있는 폴더 위치
base = "mistral"
id = "flask_evaluation_2000_fedavg_c5s3_i10_b4a1_l2048_r16a32_20241129133706"
```

```python
#허깅페이스 API
from huggingface_hub import hf_api
```

```python
#허깅페이스에 리포지토리 만들기
hf_api.create_repo(
    repo_id=f"prayslaks/{base}_{id}",
    token=write_token
)
```

```python
#모델 업로드 (repo_id는 Hugging Face에서 생성한 모델 저장소 이름)
#1부터 10라운드까지의 어댑터 모델을 한꺼번에 리포지토리로 업로드한다
#따로따로 업로드할 경우 리포지토리가 너무 많아져서 관리가 힘들기 때문
hf_api.upload_folder(
    folder_path=f"/content/drive/MyDrive/OpenFedFLASK/OpenFedLLM/output/{base}_{id}",
    repo_id=f"prayslaks/{base}_{id}",
    token=write_token
)
```
