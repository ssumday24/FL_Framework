```python
from datasets import Dataset
dataset = Dataset.from_dict({"question_id": [item["question_id"] for item in data],
                             "text": [item["text"] for item in data],
                             "task": [item["task"] for item in data],
                             "answer": [item["answer"] for item in data]})
```
